# Basic Plotting in Matplotlib

## 1. Line Plot — `plt.plot()`
A line plot is used to represent data points connected by straight lines.

Syntax: `plt.plot(x, y)`
- `x` → values on X-axis  
- `y` → values on Y-axis  

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.plot(x, y)

plt.show()
````

## 2. Plotting Multiple Lines
Multiple datasets can be plotted on the same graph using multiple `plt.plot()` calls.

- Useful for comparison between datasets

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y1 = [10, 20, 25, 30]
y2 = [5, 15, 20, 25]

plt.plot(x, y1)
plt.plot(x, y2)

plt.show()
````

## 3. Title and Axis Labels
To make plots meaningful, we add:

- Title → `plt.title()`
- X-axis label → `plt.xlabel()`
- Y-axis label → `plt.ylabel()`

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.plot(x, y)

plt.title("Growth Trend")
plt.xlabel("Time")
plt.ylabel("Value")

plt.show()
````

## 4. Legend — `plt.legend()`
Legend helps identify different lines in a graph.

- Requires `label` inside `plt.plot()`

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]

plt.plot(x, [10, 20, 30, 40], label="Product A")
plt.plot(x, [5, 15, 25, 35], label="Product B")

plt.legend()

plt.show()
````

## 5. Grid — `plt.grid()`
Grid improves readability of the plot.

- Syntax: `plt.grid(True)`

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.plot(x, y)

plt.grid(True)

plt.show()
````

### Full Example-

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]

plt.plot(x, [10, 20, 30, 40], label="Line A")
plt.plot(x, [5, 15, 25, 35], label="Line B")

plt.title("Comparison Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")

plt.legend()
plt.grid(True)

plt.show()
````

## Object-Oriented (OO) Interface in Matplotlib (with Subplots)

Matplotlib provides two approaches for plotting:

1. Pyplot (state-based) → `plt.plot()`
2. Object-Oriented (OO) Interface → `fig, ax = plt.subplots()`

The OO interface is the recommended and professional way for creating plots, especially when working with multiple visualizations.

**OO Interface is used to-**  

- Explicit control over figure and axes
- Avoids global state confusion
- Scales well for complex plots
- Essential for multiple subplots and dashboards

**Core Concept**

    fig, ax = plt.subplots()

- `fig` → Figure (entire canvas)
- `ax` → Axes (individual plot area)

````python
import matplotlib.pyplot as plt

x = [1, 2, 3]
y = [4, 5, 6]

fig, ax = plt.subplots()

ax.plot(x, y)
ax.set_title("Basic OO Plot")

plt.show()
````

## ax.plot() vs plt.plot()

**Pyplot (state-based)**

    plt.plot(x, y)
    plt.title("Plot")

- Uses global state
- Not suitable for multiple plots

**OO Interface**

    ax.plot(x, y)
    ax.set_title("Plot")

- Works on specific axes
- Clean and scalable

## Subplots in OO Interface

Subplots allow multiple plots in a single figure.

### Basic Syntax:

    fig, ax = plt.subplots(rows, cols)

## Types of Subplot Layouts

### 1. Vertical Layout (Multiple Rows)

    plt.subplots(n, 1)

- Plots stacked vertically
- Access using: ax[i]

````python
import matplotlib.pyplot as plt

x = [1, 2, 3]

fig, ax = plt.subplots(3, 1)

ax[0].plot(x, [1, 4, 9])
ax[0].set_title("Plot 1")

ax[1].plot(x, [2, 5, 8])
ax[1].set_title("Plot 2")

ax[2].plot(x, [3, 6, 9])
ax[2].set_title("Plot 3")

plt.tight_layout()
plt.show()
````

### 2. Horizontal Layout (Multiple Columns)

    plt.subplots(1, n)

- Plots side by side
- Access using: ax[i]

````python
import matplotlib.pyplot as plt

x = [1, 2, 3]

fig, ax = plt.subplots(1, 3)

ax[0].plot(x, [1, 2, 3])
ax[0].set_title("Plot 1")

ax[1].plot(x, [2, 4, 6])
ax[1].set_title("Plot 2")

ax[2].plot(x, [3, 6, 9])
ax[2].set_title("Plot 3")

plt.tight_layout()
plt.show()
````

### 3. Grid Layout (Rows × Columns)

    plt.subplots(r, c)

- Matrix of plots
- Access using: ax[i, j]

````python
import matplotlib.pyplot as plt

x = [1, 2, 3]

fig, ax = plt.subplots(2, 2)

ax[0, 0].plot(x, [1, 4, 9])
ax[0, 0].set_title("Top Left")

ax[0, 1].plot(x, [2, 5, 8])
ax[0, 1].set_title("Top Right")

ax[1, 0].plot(x, [3, 6, 9])
ax[1, 0].set_title("Bottom Left")

ax[1, 1].plot(x, [4, 7, 10])
ax[1, 1].set_title("Bottom Right")

plt.tight_layout()
plt.show()
````

### Important Concepts

- Each subplot is an independent Axes object
- Use indexing to control each plot
- Use plt.tight_layout() to adjust spacing
- Use sharex / sharey to synchronize axes

### Key Advantage

OO interface + subplots =

- Professional visualization
- Multi-plot comparison
- Dashboard-style layouts
- Real-world data analysis capability

**Example for Shared Axes-**

````python
import matplotlib.pyplot as plt

x = [1, 2, 3]

fig, ax = plt.subplots(2, 1, sharex=True)

ax[0].plot(x, [1, 4, 9])
ax[0].set_title("Shared X - Plot 1")

ax[1].plot(x, [2, 5, 8])
ax[1].set_title("Shared X - Plot 2")

plt.tight_layout()
plt.show()
````

# Axis & Scale Control in Matplotlib

Axis and scale control helps in improving the readability and accuracy of plots by controlling how data is displayed.


## 1. Axis Limits (`xlim`, `ylim`)

Axis limits define the visible range of the plot.

**Functions:**
- `plt.xlim(min, max)`
- `plt.ylim(min, max)`

**Use Cases:**
- Zoom into a specific region
- Maintain consistent scale across plots

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

plt.plot(x, y)

plt.xlim(0, 6)
plt.ylim(0, 60)

plt.title("Axis Limits Example")
plt.show()
````

## 2. Ticks (`xticks`, `yticks`)

Ticks represent the scale markings on axes.

**Functions:**
- `plt.xticks(positions, labels)`
- `plt.yticks(positions, labels)`

**Use Cases:**
- Customize axis labels
- Improve readability
- Handle categorical data

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 30, 40]

plt.plot(x, y)

plt.xticks([1, 2, 3, 4], ['One', 'Two', 'Three', 'Four'])
plt.yticks([10, 20, 30, 40])

plt.title("Ticks Customization Example")
plt.show()
````

## 3. Log Scale (`xscale`, `yscale`)

Logarithmic scale is used when data spans large ranges.

**Functions:**
- `plt.xscale('log')`
- `plt.yscale('log')`

**Use Cases:**
- Exponential growth
- Scientific and financial data

````python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(1, 100, 100)
y = x**2

plt.plot(x, y)

plt.yscale('log')

plt.title("Log Scale Example")
plt.show()
````

## 4. Axis Inversion

Axis inversion reverses the direction of an axis.

**Functions:**
- `plt.gca().invert_xaxis()`
- `plt.gca().invert_yaxis()`

**Use Cases:**
- Rankings (higher values at top)
- Depth or coordinate systems


````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 30, 40]

plt.plot(x, y)

plt.gca().invert_yaxis()

plt.title("Axis Inversion Example")
plt.show()
````

# Text & Annotation in Matplotlib

##  1. Adding Text (`plt.text()`)

Used to place text at a specific coordinate `(x, y)` on the plot.

### Key Features:
- Simple labeling
- No arrows
- Fully customizable (color, size, alignment)

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y)

plt.text(2, 20, "Important Point", fontsize=10, color='blue')

plt.title("Using plt.text()")
plt.show()
````

##  2. Annotations (`plt.annotate()`)

Used to highlight points with **text + arrows**.

### Key Features:
- Points to exact location
- Better for explanations
- Supports arrow styling

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y)

plt.annotate("Peak Value",
             xy=(4, 25),
             xytext=(3, 30),
             arrowprops=dict(facecolor='black', shrink=0.05))

plt.title("Using plt.annotate()")
plt.show()
````

##  3. Highlighting Important Points

Techniques:

# Working with Images in Matplotlib

Matplotlib provides functionality to display image data using the `imshow()` function. It is commonly used for visualizing 2D arrays such as grayscale images, heatmaps, and pixel-based datasets.

## 1. Displaying Images using `plt.imshow()`

- `imshow()` stands for **image show**.
- It displays data as an image.
- Accepts:
  - 2D array → Grayscale image
  - 3D array → RGB image

````python
import matplotlib.pyplot as plt
import numpy as np

# Create a random 2D array
img = np.random.rand(10, 10)

plt.imshow(img)
plt.title("Basic Image Display")
plt.show()
````

## 2. Colormaps (`cmap`)

A **colormap** maps numerical values to colors.

### Types of Colormaps:
- **Sequential** → `viridis`, `plasma` (for gradual data)
- **Diverging** → `coolwarm`, `bwr` (for positive & negative values)
- **Qualitative** → `tab10`, `Set1` (for categories)

````python
plt.imshow(img, cmap='hot')
plt.title("Image with Hot Colormap")
plt.show()
````

## 3. Colorbar (`plt.colorbar()`)

- A colorbar shows the relationship between **data values and colors**.
- It improves interpretability of the image.

````python
plt.imshow(img, cmap='viridis')
plt.colorbar()
plt.title("Image with Colorbar")
plt.show()
````

**Combined Example (Colormap + Colorbar)**

````python
data = np.random.rand(20, 20)

plt.imshow(data, cmap='coolwarm')
plt.colorbar()
plt.title("Heatmap Example")
plt.show()
````

## Important Parameters of `imshow()`

- `cmap` → Defines color mapping
- `vmin`, `vmax` → Control color scaling
- `interpolation` → Pixel smoothing
- `aspect` → Aspect ratio control

````python
plt.imshow(data, cmap='plasma', vmin=0, vmax=1, interpolation='nearest')
plt.colorbar()
plt.title("Customized Image Display")
plt.show()
````

# Advanced Features in Matplotlib

## 1. Twin Axes (`ax.twinx()`)

### Concept:
- Used to plot **two different datasets with different scales**
- Shares the same x-axis
- Creates a second y-axis on the right side

### Use Case:
- Temperature vs Humidity
- Price vs Volume

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
temp = [30, 32, 35, 33, 31]
humidity = [60, 65, 70, 68, 66]

fig, ax1 = plt.subplots()

ax1.plot(x, temp, color='red')
ax1.set_ylabel("Temperature (°C)", color='red')

ax2 = ax1.twinx()
ax2.plot(x, humidity, color='blue')
ax2.set_ylabel("Humidity (%)", color='blue')

plt.title("Temperature vs Humidity")
plt.show()
````

##  2. Secondary Axis (`secondary_xaxis` / `secondary_yaxis`)

### Concept:
- Used to represent **the same data in different units**
- Based on transformation functions

### Difference from Twin Axes:
- Twin axes → different datasets
- Secondary axis → same dataset, different units

### Use Case:
- Celsius ↔ Fahrenheit conversion

````python
import numpy as np
import matplotlib.pyplot as plt

def celsius_to_fahrenheit(x):
    return x * 9/5 + 32

def fahrenheit_to_celsius(x):
    return (x - 32) * 5/9

x = np.linspace(0, 100, 10)

fig, ax = plt.subplots()
ax.plot(x, x)

sec_ax = ax.secondary_yaxis('right',
            functions=(celsius_to_fahrenheit, fahrenheit_to_celsius))

ax.set_ylabel("Celsius")
sec_ax.set_ylabel("Fahrenheit")

plt.title("Temperature Conversion")
plt.show()
````

##  3. Area Plots (`fill_between()`)

### Concept:
- Fills the area under a curve or between two curves

### Use Case:
- Highlighting ranges
- Confidence intervals
- Profit/Loss visualization

**Area Plot**

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 1, 5, 3]

plt.plot(x, y)
plt.fill_between(x, y, color='skyblue', alpha=0.4)

plt.title("Area Plot")
plt.show()
````

**Area Between Two Lines**

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [2, 4, 6, 8, 10]
y2 = [1, 3, 5, 7, 9]

plt.fill_between(x, y1, y2, color='green', alpha=0.3)

plt.title("Area Between Two Curves")
plt.show()
````

##  4. Error Bars (`errorbar()`)

### Concept:
- Represents **uncertainty or variation** in data

### Parameters:
- `yerr` → vertical error
- `xerr` → horizontal error
- `capsize` → cap width

### Use Case:
- Scientific data
- Statistical analysis (mean ± deviation)

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 18]
error = [2, 3, 1, 4, 2]

plt.errorbar(x, y, yerr=error, fmt='o', capsize=5)

plt.title("Error Bar Plot")
plt.show()
````

# Saving & Exporting in Matplotlib

Saving plots is essential for using visualizations in reports, presentations, or web applications. Matplotlib provides the `plt.savefig()` function to export figures in various formats with customizable resolution and layout.

##  plt.savefig()

The `savefig()` function saves the current figure to a file.

### Key Points:
- Must be called **before `plt.show()`**
- Supports multiple formats like PNG, PDF, SVG, etc.
- Can control resolution using DPI
- Can adjust layout to prevent clipping

````python
import matplotlib.pyplot as plt

x = [1, 2, 3]
y = [2, 4, 6]

plt.plot(x, y)
plt.title("Simple Plot")

plt.savefig("plot.png")
plt.show()
````

## 🔹 File Formats

Matplotlib supports different output formats:

- **PNG** → Raster format, best for screens and presentations  
- **JPG** → Lossy format, not ideal for plots  
- **PDF** → Vector format, ideal for academic papers  
- **SVG** → Scalable vector format, useful for web  
- **EPS** → Used in publishing and LaTeX  

````python
import matplotlib.pyplot as plt

x = [1, 2, 3]
y = [3, 6, 9]

plt.plot(x, y)
plt.title("Multiple Formats")

plt.savefig("plot.png")
plt.savefig("plot.pdf")
plt.savefig("plot.svg")

plt.show()
````

## 🔹 DPI (Dots Per Inch)

DPI determines the resolution of the saved image.

- 72 DPI → Screen display  
- 150 DPI → Medium quality  
- 300 DPI → Print quality  
- 600 DPI → High-quality publication  

````python
import matplotlib.pyplot as plt

x = [1, 2, 3]
y = [1, 4, 9]

plt.plot(x, y)
plt.title("High Resolution Plot")

plt.savefig("high_quality.png", dpi=300)
plt.show()
````

#  Performance & Best Practices in Matplotlib

##  Clearing Figures

Matplotlib retains figures in memory unless explicitly cleared or closed.

- `plt.clf()` → Clears the current figure (keeps window open)
- `plt.cla()` → Clears only the current axes
- `plt.close()` → Closes the figure completely (frees memory)

-> Important when generating multiple plots in loops.

````python
import matplotlib.pyplot as plt

for i in range(3):
    plt.plot([1, 2, 3], [i, i+1, i+2])
    plt.title(f"Plot {i}")
    plt.savefig(f"plot_{i}.png")
    
    plt.clf()  # Clear figure for next iteration
````

##  Efficient Plotting




- Use NumPy arrays instead of Python lists for better performance
- Avoid redundant plotting

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 1000)
y = np.sin(x)

plt.plot(x, y)
plt.title("Efficient Plot with NumPy")
plt.show()
````

- Prefer Object-Oriented (OO) interface (`fig, ax`) for better control and scalability

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)
y = np.cos(x)

fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_title("Using OO Interface")
plt.show()
````

##  Avoiding Visual Clutter

Cluttered plots reduce readability.

### Common Issues:
- Too many lines or markers
- Overlapping labels
- Excessive colors

### Solutions:
- Use fewer elements
- Apply transparency (`alpha`)
- Adjust figure size
- Rotate tick labels if needed
- Use grid lines carefully

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.random.rand(100)
y = np.random.rand(100)

plt.scatter(x, y, alpha=0.5)
plt.title("Reduced Clutter with Transparency")
plt.show()
````

##  Optimizing Large Data

- Downsample data (plot fewer points)
- Avoid markers in large datasets
- Prefer line plots over scatter plots for dense data


````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 100, 10000)
y = np.sin(x)

plt.plot(x[::10], y[::10])  # Plot every 10th point
plt.title("Downsampled Plot")
plt.show()
````

##  Reusing Figures

Instead of creating new figures repeatedly, reuse the same axes when possible to improve performance.



````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)

fig, ax = plt.subplots()

for i in range(3):
    ax.plot(x, x + i, label=f"Line {i}")

ax.legend()
plt.show()
````

## Layout Adjustment

- `plt.tight_layout()` helps prevent overlapping elements

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))
plt.plot([1, 2, 3], [4, 5, 6])
plt.title("Tight Layout Example")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.tight_layout()
plt.show()
````

#  Real-World Applications of Matplotlib

Matplotlib is widely used for analyzing and visualizing real-world datasets. This section covers four major practical applications:


## 1. Time Series Visualization

Time series visualization is used to display data points over time.

### Use Cases:
- Stock market trends
- Weather data
- Student performance over time

### Key Features:
- Line plots (`plt.plot()`)
- Date handling using pandas
- Rotating axis labels for clarity

````python
import matplotlib.pyplot as plt
import pandas as pd

dates = pd.date_range(start='2025-01-01', periods=10)
values = [10, 12, 15, 14, 18, 20, 22, 21, 19, 23]

plt.plot(dates, values, marker='o')
plt.title("Time Series Data")
plt.xlabel("Date")
plt.ylabel("Value")
plt.xticks(rotation=45)
plt.grid()

plt.show()
````

## 2. Comparison Charts

Comparison charts help in comparing multiple categories or datasets.

### Types:
- Bar charts
- Grouped bar charts
- Line comparisons

### Use Cases:
- Comparing student scores
- Sales comparison across regions
- Algorithm performance comparison

````python
import matplotlib.pyplot as plt
import numpy as np

subjects = ['Math', 'Science', 'English']
student1 = [85, 90, 88]
student2 = [80, 85, 92]

x = np.arange(len(subjects))
width = 0.35

plt.bar(x - width/2, student1, width, label='Student 1')
plt.bar(x + width/2, student2, width, label='Student 2')

plt.xticks(x, subjects)
plt.title("Student Performance Comparison")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.legend()

plt.show()
````

## 3. Distribution Analysis

Distribution analysis shows how data is spread.

### Types:
- Histogram (`plt.hist()`)
- Box plot (`plt.boxplot()`)

### Use Cases:
- Exam score distribution
- Data variability analysis
- Outlier detection

````python
import matplotlib.pyplot as plt
import numpy as np

data = np.random.normal(50, 10, 100)

plt.hist(data, bins=10, edgecolor='black')
plt.title("Histogram of Data Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
````

````python
import matplotlib.pyplot as plt

plt.boxplot(data)
plt.title("Box Plot for Outlier Detection")

plt.show()
````

## 4. Dashboard-Style Plots

Dashboard-style plots combine multiple plots into one figure.

### Features:
- Multiple subplots
- Grid layouts
- Combined insights

### Use Cases:
- Business dashboards
- Academic reports
- Data storytelling

````python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(2, 2, figsize=(10, 8))

# Line plot
ax[0, 0].plot([1, 2, 3], [4, 5, 6])
ax[0, 0].set_title("Line Plot")

# Bar chart
ax[0, 1].bar(['A', 'B', 'C'], [5, 7, 3])
ax[0, 1].set_title("Bar Chart")

# Histogram
ax[1, 0].hist([1, 2, 2, 3, 3, 3])
ax[1, 0].set_title("Histogram")

# Pie chart
ax[1, 1].pie([30, 40, 30], labels=['A', 'B', 'C'])
ax[1, 1].set_title("Pie Chart")

plt.tight_layout()
plt.show()
````
